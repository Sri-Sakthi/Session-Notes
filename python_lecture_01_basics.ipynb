## 1. What is Python?

Python is a **high-level, dynamically typed, multi-paradigm programming language** that executes code line by line.

- **High-level**: easy to read, close to human language. Much less code than Java.
- **Dynamically typed**: you do NOT declare a data type for variables. Python figures it out automatically.
- **Multi-paradigm**: works for data science, ML, AI, web apps, automation, backend APIs.
- **Interpreted**: code runs line by line — if line 3 fails, lines 1 and 2 already ran.

**Python vs SQL** (from class):
- SQL = structured query language. Deals with databases only. Queries data.
- Python = general purpose. Gives instructions to the whole machine. Line by line execution.
- Java needs `public class`, `static void main`, `System.out.println` just to print one line.
- Python needs just `print("Hello")`. That's it.

## 4. Variables and Variable Assignment

A **variable** is a container in memory used to store data.
Think of it as a labeled box. The variable name is the label. The value inside is the data.

When you assign a value, Python creates a memory location and the variable name points to it.
You do not store the value in the variable itself -- the variable is a **reference** to memory.

**Variable Naming Rules** (from class):
- Can start with a letter or underscore `_`
- Can contain letters, numbers, underscores
- CANNOT start with a number (`9name` is invalid)
- CANNOT use special characters like `$`, `@`, `-`
- CANNOT use Python built-in names like `print`, `list`, `int`, `str`
  (these show in green in Jupyter — that means Python already uses them)

````python
# Basic variable assignment
a = 9
b = 3
print(a + b)   # 12
````

````python
# Variable reassignment
# Initially a points to memory holding 9
a = 9
print(a)  # 9

# Now a points to memory holding 10. The 9 stays in memory until garbage collected.
a = 10
print(a)  # 10

# Reassign back to 9 -- Python reuses the existing memory for 9 (does not create new)
a = 9
print(a)  # 9
````

````python
# Valid variable names
my_name = "Santhakumar"
_age = 25
name2 = "Ramesh"
print(my_name, _age, name2)
````

## 5. Type Conversion (Type Casting)

Converting a value from one data type to another. Two types:

- **Implicit conversion**: Python does it automatically. No input from you. Avoids data loss.
  Example: int + float → Python converts int to float automatically. Result is float.

- **Explicit conversion**: You manually convert using built-in functions: `int()`, `float()`, `str()`, `bool()`.
  Needed when Python cannot automatically handle incompatible types (like str + int → error).

**SQL comparison from class**: In SQL we explicitly define column data type when creating a table (`INT`, `VARCHAR`).
In Python the type is decided dynamically at runtime based on the value assigned.

````python
# Implicit conversion -- Python handles it automatically
x = 5      # int
y = 2.0    # float
z = x + y  # Python sees int + float and converts int to float automatically
print(z)         # 7.0
print(type(z))   # float
````

````python
# Explicit conversion -- you do it manually
a = 7                    # int
# print("Age: " + a)    # ERROR: can only concatenate str, not int
print("Age: " + str(a)) # Correct: convert int to str first using str()
print(type(str(a)))      # <class 'str'>
````

````python
# int() converts string to integer
age = int(input("Enter your age: "))
print(age)
print(type(age))   # <class 'int'>
````

## 6. Dynamic Typing and Duck Typing

**Dynamic typing**: You do NOT declare the data type when creating a variable.
Python determines the type automatically at runtime based on the value assigned.
You can reassign the same variable to a completely different type and Python accepts it.

**SQL comparison from class**: In SQL, a column is declared as `INT` and can only hold integers.
In Python, a variable can hold an int, then later hold a string — no error.

**Duck Typing**: "If it walks like a duck and quacks like a duck, it is a duck."
Python does not care about what an object IS. It cares about what it CAN DO.
If an object has the method you need, Python will accept it — regardless of its class or type.

````python
# Dynamic typing in action
x = 10
print(x, type(x))   # 10 <class 'int'>

x = "hello"
print(x, type(x))   # hello <class 'str'>

x = 3.14
print(x, type(x))   # 3.14 <class 'float'>

x = True
print(x, type(x))   # True <class 'bool'>
# Same variable x. Different type each time. Python accepts it all.
````

## 7. Operators in Python

### 7.1 Mathematical (Arithmetic) Operators

````python
a = 9
b = 3

print(a + b)   # Addition       -> 12
print(a - b)   # Subtraction    -> 6
print(a * b)   # Multiplication -> 27
print(a / b)   # Division       -> 3.0  (always returns float)
print(a // b)  # Floor Division -> 3    (quotient only, drops decimal)
print(a % b)   # Modulus        -> 0    (remainder after division)
print(a ** b)  # Power          -> 729  (a to the power of b)
````

**Floor Division vs Modulus** (from class):
- `10 // 4` = **2** → quotient (how many full times 4 goes into 10)
- `10 % 4`  = **2** → remainder (what is left after dividing)
- Instructor example: 10 ÷ 4 = 2 remainder 2. `//` gives 2. `%` gives 2.

### 7.2 Comparison Operators — result is always True or False

````python
a = 10
b = 20

print(a == b)   # Equal to              -> False
print(a != b)   # Not equal to          -> True
print(a > b)    # Greater than          -> False
print(a < b)    # Less than             -> True
print(a >= b)   # Greater than or equal -> False
print(a <= b)   # Less than or equal    -> True
````

### 7.3 Logical Operators — combine conditions

````python
# AND -- both conditions must be True
age = 25
print(age > 18 and age < 60)   # True -- both are true

# OR -- at least one condition must be True
age = 70
print(age < 18 or age > 60)    # True -- second condition is true

# NOT -- reverses/negates the result
is_logged_in = True
print(not is_logged_in)         # False -- reverses True to False
````

**SQL comparison from class**: AND, OR, NOT work exactly the same in SQL WHERE clauses.
The logic is identical. Just the syntax differs slightly.

### 7.4 Assignment Operators — shorthand for updating variables

````python
x = 10
print(x)        # 10

x += 5          # same as: x = x + 5
print(x)        # 15

x -= 3          # same as: x = x - 3
print(x)        # 12

x *= 2          # same as: x = x * 2
print(x)        # 24

x /= 4          # same as: x = x / 4
print(x)        # 6.0

x %= 4          # same as: x = x % 4
print(x)        # 2.0

x **= 3         # same as: x = x ** 3
print(x)        # 8.0
````

**From class**: `x += 5` means `x = x + 5`. The `+` on the LEFT of `=` is an operation.
The `+` on the RIGHT of `=` is a positive sign. Position matters.

## 9. F-String and .format()

Both are ways to **embed variables inside a string** cleanly.
Before these existed, you had to use `+` concatenation which is messy for long strings.

````python
# Old way: string concatenation with +
name = "Sakthi"
age  = 23
print("Hello " + name + ", you are " + str(age) + " years old.")
# Messy and error-prone. Must convert non-strings with str().
````

````python
# F-String: put f before the quote, use {} for variables
name = "Sakthi"
age  = 23
print(f"Hello {name}, you are {age} years old.")
# Clean, readable, no need to convert types manually
# You can even put expressions inside {}: f"Result: {2 + 3}" 
````

````python
# .format() method: older way, still valid
language = "Python"
subject  = "Data Science"
print("I love {}. My subject is {}.".format(language, subject))
# {} are placeholders. .format() fills them in order.
````

````python
# F-string with expressions (bonus)
a = 50
born_year = 1978
print(f"I am {a} years old, born in {born_year}.")
# Change a or born_year above -- print updates automatically. No hardcoding.
````

## 11. Branching — if / elif / else

Branching means making decisions in code based on conditions.
**SQL equivalent from class**: This is the same as the `CASE WHEN ... THEN ... ELSE` statement in SQL.

Structure:
```
if condition1:
    # do this if condition1 is True
elif condition2:
    # do this if condition1 failed but condition2 is True
else:
    # do this if ALL conditions above failed
```
Python stops checking after the FIRST True condition. Order matters.
Indentation (4 spaces) is how Python knows what belongs inside each block.

````python
i = 12

if i < 3:
    print("Less than 3")
elif i < 5:
    print("Between 3 and 5")
elif i < 12:
    print("Between 5 and 12")
else:
    print("12 or greater")   # This runs because all above conditions failed
````

````python
# Marks grading example
marks = 82

if marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")    # This runs -- 82 >= 75 is True
elif marks >= 60:
    print("Grade C")
else:
    print("Fail")
````

````python
# Wrong order -- common mistake
# Python stops at the FIRST True condition
marks = 95

if marks >= 60:
    print("Grade C")    # This runs even though marks = 95 should be Grade A
elif marks >= 75:
    print("Grade B")    # Never reached
elif marks >= 90:
    print("Grade A")    # Never reached

# Always write most specific (highest) condition first
````

### 11.1 Nested if — if inside another if

````python
# Nested if: check a condition inside another condition
age = 25
has_license = True

if age >= 18:
    # We are inside the first if now
    if has_license:
        print("Can Drive")     # Both conditions are True
    else:
        print("License Required")  # Age is fine but no license
else:
    print("Too Young")  # Age failed, never checks license
````

````python
# Class example of nested if
i = 2

if i < 3:
    print("Value is less than 3")
    if i < 2:
        print("Value is also less than 2")  # Only runs if i < 3 AND i < 2
````
